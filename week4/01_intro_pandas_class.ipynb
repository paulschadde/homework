## Intro to numpy
### List → NumPy array

```python
import numpy as np

scores = [70, 80, 90]
arr = np.array(scores)

print(scores)
print(arr)
print(type(scores))   # list
print(type(arr))      # numpy.ndarray
```

```python
# type = kind of object; dtype = type of values inside
print(arr.dtype)      # integers
print(arr.shape)      # (3,) = 3 values in one dimension --> Vector
print(arr[0])         # 70: first value, position 0

print(scores * 2)     # repeats the list
print(arr * 2)        # doubles each value
```

## Intro to pandas
### List or array → pandas Series
A Series is one-dimensional data with index labels.

```python
import pandas as pd

s = pd.Series(scores)             # from a Python list
s_from_array = pd.Series(arr)      # from a NumPy array

print(s)
print(s_from_array)               # same values and default index
print(type(s))                    # pandas.Series
print(s.dtype)                    # type of values
print(s.shape)                    # (3,)
print(s.index)                    # default labels: 0, 1, 2
```

```python
# We can choose our own index labels
named_scores = pd.Series(arr, index=["Ana", "Ben", "Cara"])
print(named_scores)
print(named_scores.loc["Ana"])    # 70: select by label
```

### Same numeric operations

```python
# Apply an operation to every value, without a loop
print(arr + 10)
print(s + 10)        # Series keeps its index labels

# These expressions return new results
print(arr)
print(s)             # original scores are unchanged
```

```python
# NumPy array and Series: the same method names
print(arr.sum(), s.sum())       # 240, 240
print(arr.mean(), s.mean())     # 80.0, 80.0
print(arr.min(), s.min())       # 70, 70
print(arr.max(), s.max())       # 90, 90
```

### Dictionary → DataFrame
A DataFrame is a two-dimensional table with row and column labels.

```python
# Dictionary keys become column names; lists must have equal length
students = {
    "name": ["Ana", "Ben", "Cara"],
    "score": scores,
}
df = pd.DataFrame(students)

print(type(students))  # dict
print(type(df))        # pandas.DataFrame
print(df)
```

```python
print(df.shape)       # (3, 2) = 3 rows, 2 columns
print(df.columns)     # column names
print(df.index)       # row labels
print(df.dtypes)      # one dtype per column; text and numbers can differ
```

### Select a column: Series or DataFrame?

```python
score_series = df["score"]       # one column name → Series
score_frame = df[["score"]]      # list of column names → DataFrame

print(score_series)
print(score_frame)
print(type(score_series), score_series.shape)  # Series, (3,)
print(type(score_frame), score_frame.shape)    # DataFrame, (3, 1)
```

```python
# A selected numeric column works just like our earlier Series
print(df["score"].mean())         # 80.0
print(df["score"] + 10)           # add 10 to each score
print(df[["score"]].mean())       # Series: one mean per selected column
```
